```python
import io
import pandas as pd
import subprocess as sbp
import matplotlib.pyplot as plt
import seaborn as sns


def get_configurations(n, k):
    r = sbp.getoutput(f"ewens-configurations {n} {k}")
    return pd.read_csv(
        io.StringIO(r), header=None, names=["n", "k", "c", "pe", "pw"]
    ).sort_values(["pe", "pw"])
```

```python
d1 = get_configurations(16, 7)
d2 = get_configurations(27, 14)
d3 = get_configurations(41, 23)
d4 = get_configurations(68, 31)
```

Here, `pe` is the probability of unordered configuration under the Ewens model, `pw` is sample homozygosity.

```python
d1.sort_values(["pe"])
```

```python
sns.histplot(d1, x="pe", log_scale=True)
```

```python
d2.sort_values(["pe"])
```

```python
sns.histplot(d2, x="pe", log_scale=True)
```

```python
sns.scatterplot(d2, x="pw", y="pe", linewidth=0, alpha=0.4)
plt.xscale("log")
plt.yscale("log")
plt.ylabel("Probability of unordered configuration under the Ewens model")
plt.xlabel("Homozygosity")
```
